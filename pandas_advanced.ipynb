```python
import pandas as pd

orders = pd.DataFrame({
    'order_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'customer_id': [1, 2, 1, 3, 4, 2, 5],
    'category': ['Electronics', 'Books', 'Electronics',
                 'Home', 'Books', 'Home', 'Electronics'],
    'amount': [1200, 25, 900, 150, 30, 220, 1600],
    'order_date': ['2026-04-01', '2026-04-01', '2026-04-02',
                   '2026-04-02', '2026-04-03', '2026-04-03', '2026-04-04']
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'city': ['Toronto', 'Seoul', 'Toronto', 'Vancouver', 'Seoul'],
    'segment': ['Business', 'Student', 'Business', 'Student', 'Business']
})
```

```python
print(orders.head())
print(orders.info())
print(orders.describe())
```

```python
print(customers.head())
print(customers.info())
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
print(orders.dtypes)
```

```python
orders['tax'] = orders['amount'] * 0.13
orders['final_amount'] = orders['amount'] + orders['tax']
orders['is_large_order'] = orders['amount'] >= 500

print(orders.head())
```

```python
category_summary = orders.groupby('category')['amount'].agg(
    ['count', 'sum', 'mean', 'max']
)

print(category_summary)
```

```python
city_category = orders.groupby(['category', 'customer_id'])['amount'].sum()
print(city_category)
```

```python
segment_summary = orders.groupby('customer_id')['amount'].sum()
print(segment_summary)
```

```python
full_data = orders.merge(customers, on='customer_id', how='left')
print(full_data)
```

```python
segment_revenue = full_data.groupby('segment')['amount'].sum()
segment_avg = full_data.groupby('segment')['amount'].mean()

print(segment_revenue)
print(segment_avg)
```

```python
city_summary = full_data.groupby('city')['amount'].agg(
    ['count', 'sum', 'mean']
).sort_values('sum', ascending=False)

print(city_summary)
```

### Exercise 2.

GroupBy with multiple columns, Aggregation (15분)

- Using full_data, group the data by both city and category.
- Calculate the sum and mean of the amount for each city-category pair.
- Identify which city has the highest order amount in the 'Electronics' category.

```python
city_category_summary = full_data.groupby(['city', 'category'])['amount'].agg(['sum', 'mean'])
print(city_category_summary)

electronics = full_data[full_data['category'] == 'Electronics']

electronics_city = electronics.groupby('city')['amount'].sum()
print(electronics_city)

print(electronics_city.idxmax())
```

### Mini Exercise 3.

```python
def label_order(x):
    if x >= 1000:
        return 'High'
    elif x >= 100:
        return 'Medium'
    else:
        return 'Low'

orders['discount_band'] = orders['amount'].apply(label_order)
print(orders[['amount', 'discount_band']])
```
