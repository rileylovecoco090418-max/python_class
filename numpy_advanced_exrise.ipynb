```python
import numpy as np

# rows = days, columns = rooms
energy = np.array([
    [30, 45, 22, 60, 35],
    [32, 47, 21, 58, 36],
    [31, 50, 25, 65, 38],
    [29, 43, 20, 55, 34],
    [35, 52, 26, 70, 40],
    [40, 60, 30, 80, 45],
    [38, 58, 28, 78, 44]
])

print(energy.shape)   # (7, 5)
```

```python
# total usage per room
print(energy.sum(axis=0))

# total usage per day
print(energy.sum(axis=1))

# average usage per room
print(energy.mean(axis=0))
```

## Finding Patterns Fast

```python
room_totals = energy.sum(axis=0)
day_totals = energy.sum(axis=1)

most_expensive_room = np.argmax(room_totals)
highest_usage_day = np.argmax(day_totals)

print(room_totals)
print(day_totals)
print(most_expensive_room)
print(highest_usage_day)
```

```python

# Find all readings above 50 kWh
high_usage = energy > 50
print(high_usage)

# Extract only those values
print(energy[high_usage])

# Count how many high-usage readings exist
print(np.sum(high_usage))

```

```python
budget = np.array([35, 50, 25, 65, 40])

over_budget = energy > budget
print(over_budget)

```

```python
over_budget = energy > budget

# Count over-budget days per room
print(over_budget.sum(axis=0))

# Count over-budget rooms per day
print(over_budget.sum(axis=1))

# Replace normal values with 0, keep only over-budget readings
flagged = np.where(over_budget, energy, 0)
print(flagged)



```

```python
room_means = energy.mean(axis=0)
normalized = energy / room_means

print(room_means)
print(normalized.round(2))
```

```python
day_totals = energy.sum(axis=1)

mean_total = day_totals.mean()
std_total = day_totals.std()

z_scores = (day_totals - mean_total) / std_total
print(z_scores.round(2))


unusual_days = np.abs(z_scores) > 1.2
print(unusual_days)
```
