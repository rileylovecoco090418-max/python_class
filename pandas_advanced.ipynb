```python
import pandas as pd

orders = pd.DataFrame({
    'order_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'customer_id': [1, 2, 1, 3, 4, 2, 5],
    'category': ['Electronics', 'Books', 'Electronics',
                 'Home', 'Books', 'Home', 'Electronics'],
    'amount': [1200, 25, 900, 150, 30, 220, 1600],
    'order_date': ['2026-04-01', '2026-04-01', '2026-04-02',
                   '2026-04-02', '2026-04-03', '2026-04-03', '2026-04-04']
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'city': ['Toronto', 'Seoul', 'Toronto', 'Vancouver', 'Seoul'],
    'segment': ['Business', 'Student', 'Business', 'Student', 'Business']
})
```

```python
print(orders.head())
print(orders.info())
print(orders.describe())
```

```python
print(customers.head())
print(customers.info())
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
print(orders.dtypes)
```

```python
orders['tax'] = orders['amount'] * 0.13
orders['final_amount'] = orders['amount'] + orders['tax']
orders['is_large_order'] = orders['amount'] >= 500

print(orders.head())
```

```python
category_summary = orders.groupby('category')['amount'].agg(
    ['count', 'sum', 'mean', 'max']
)

print(category_summary)
```

```python
city_category = orders.groupby(['category', 'customer_id'])['amount'].sum()
print(city_category)
```

```python
segment_summary = orders.groupby('customer_id')['amount'].sum()
print(segment_summary)
```
